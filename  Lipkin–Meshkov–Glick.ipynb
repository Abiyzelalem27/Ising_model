# Ground-State Quantum Phase Transition in the LMG Model

## 1. Hamiltonian

The form of the Lipkin–Meshkov–Glick model considered here is

$$
H=-\frac{J}{N}S_z^2-\Omega S_x.
$$

The collective spin operators are

$$
S_\alpha
=
\frac{1}{2}
\sum_{i=1}^{N}\sigma_\alpha^{(i)},
\qquad
\alpha\in\{x,y,z\}.
$$

### Meaning of the symbols

- $H$ is the Hamiltonian, which represents the total energy of the spin system.
- $N$ is the total number of spin-$\frac12$ particles or qubits.
- $J$ is the strength of the collective spin–spin interaction.
- $\Omega$ is the strength of the external transverse field.
- $S_z$ is the total collective spin in the $z$-direction.
- $S_z^2$ represents the collective interaction between the spins.
- $S_x$ is the total collective spin in the $x$-direction.
- $\sigma_\alpha^{(i)}$ is a Pauli operator acting on spin $i$ in the $\alpha$-direction.

The factor $1/N$ ensures that the total energy grows approximately linearly with $N$. Consequently, the energy per spin remains finite as $N\rightarrow\infty$.

---

```python
N = 10

# First Hamiltonian
J_1 = 4
Omega_1 = 1

H_1, Sx, Sy, Sz, S_plus, S_minus =  im.operators.LMG_hamiltonian(N, J_1, Omega_1)

print("First Hamiltonian:")
print("Omega/J =", Omega_1 / J_1)
print(np.round(H_1.real, 3))
print()


# Second Hamiltonian
J_2 = 2
Omega_2 = 6

H_2, Sx, Sy, Sz, S_plus, S_minus=  im.operators.LMG_hamiltonian(N, J_2, Omega_2)

print("Second Hamiltonian:")
print("Omega/J =", Omega_2 / J_2)
print(np.round(H_2.real, 3))
```

## 2. Competition Between the Hamiltonian Terms

The Hamiltonian contains two competing contributions:

$$
H_{\mathrm{int}}
=
-\frac{J}{N}S_z^2
$$

and

$$
H_{\mathrm{field}}
=
-\Omega S_x.
$$

The interaction term $H_{\mathrm{int}}$ favours collective spin alignment along the $z$-direction.

The transverse-field term $H_{\mathrm{field}}$ favours collective spin alignment along the $x$-direction.

The behavior of the system is therefore controlled by the dimensionless ratio

$$
\lambda=\frac{\Omega}{J}.
$$

This ratio compares the strength of the transverse field with the collective interaction.

---

## 3. Ferromagnetic Regime

When

$$
\Omega\ll J
\qquad\text{or}\qquad
\frac{\Omega}{J}\ll1,
$$

the interaction term dominates:

$$
H\approx-\frac{J}{N}S_z^2.
$$

Since this term has a negative sign, the energy is minimized by maximizing $S_z^2$.

The maximum possible values of $S_z$ are

$$
S_z=\pm\frac{N}{2}.
$$

These values correspond to the two fully aligned states

$$
|00\cdots0\rangle
$$

and

$$
|11\cdots1\rangle.
$$

Thus, the system prefers all spins to point in the same $z$-direction. This behavior defines the **ferromagnetic phase**.

Using the convention

$$
\sigma_z|0\rangle=|0\rangle,
\qquad
\sigma_z|1\rangle=-|1\rangle,
$$

we have

$$
S_z|00\cdots0\rangle
=
\frac{N}{2}|00\cdots0\rangle
$$

and

$$
S_z|11\cdots1\rangle
=
-\frac{N}{2}|11\cdots1\rangle.
$$

Both states have the same value of $S_z^2$:

$$
S_z^2|00\cdots0\rangle
=
\frac{N^2}{4}|00\cdots0\rangle,
$$

$$
S_z^2|11\cdots1\rangle
=
\frac{N^2}{4}|11\cdots1\rangle.
$$

Therefore, the interaction term gives both states the same energy.

---

## 4. Spin-Flip Symmetry

The Hamiltonian is invariant under a simultaneous flip of all spins:

$$
|0\rangle\leftrightarrow|1\rangle.
$$

This is called a global $\mathbb{Z}_2$ spin-flip symmetry.

The transformation exchanges

$$
|00\cdots0\rangle
\longleftrightarrow
|11\cdots1\rangle.
$$

For a finite system with a small nonzero transverse field, the symmetry-preserving ground state is approximately

$$
|\mathrm{GS}_+\rangle
\approx
\frac{
|00\cdots0\rangle+
|11\cdots1\rangle
}{\sqrt2}.
$$

The corresponding antisymmetric state is approximately

$$
|\mathrm{GS}_-\rangle
\approx
\frac{
|00\cdots0\rangle-
|11\cdots1\rangle
}{\sqrt2}.
$$

These are Schrödinger-cat-like states.

In the thermodynamic limit $N\rightarrow\infty$, the system can select one of the two orientations:

$$
|00\cdots0\rangle
\quad\text{or}\quad
|11\cdots1\rangle.
$$

This selection is called **spontaneous symmetry breaking**.

For a finite symmetry-preserving system, the exact expectation value $\langle S_z\rangle$ may be zero because the two orientations contribute equally. A useful finite-system order parameter is therefore

$$
m_z
=
\frac{2}{N}
\sqrt{\langle S_z^2\rangle}.
$$

In the ferromagnetic regime, $m_z$ is large.

---

## 5. Paramagnetic Regime

When

$$
\Omega\gg J
\qquad\text{or}\qquad
\frac{\Omega}{J}\gg1,
$$

the transverse-field term dominates:

$$
H\approx-\Omega S_x.
$$

To minimize the energy, the collective spin aligns along the positive $x$-direction.

For one spin, the positive $x$-eigenstate is

$$
|+\rangle
=
\frac{|0\rangle+|1\rangle}{\sqrt2},
$$

and it satisfies

$$
\sigma_x|+\rangle=|+\rangle.
$$

For $N$ spins, the field-aligned state is

$$
|\mathrm{GS}\rangle
\approx
|+\rangle^{\otimes N}
=
\left(
\frac{|0\rangle+|1\rangle}{\sqrt2}
\right)^{\otimes N}.
$$

In this state,

$$
\langle S_x\rangle\approx\frac{N}{2},
$$

so the normalized $x$-magnetization is

$$
m_x
=
\frac{2}{N}\langle S_x\rangle
\approx1.
$$

The average $z$-magnetization is approximately zero:

$$
\langle S_z\rangle\approx0.
$$

This field-aligned regime is called the **paramagnetic phase**.

---


```python
parameter_sets = {
    "First system": {
        "J": 4.0,
        "Omega": 1.0
    },
    "Second system": {
        "J": 2.0,
        "Omega": 6.0
    }
}

for name, parameters in parameter_sets.items():

    J = parameters["J"]
    Omega = parameters["Omega"]
    ratio = Omega / J

    # The star collects any additional returned operators
    H, Sx, Sy, Sz, *_ = im.operators.LMG_hamiltonian(
        N, J, Omega
    )

    # Diagonalize H
    energies, eigenvectors = np.linalg.eigh(H)

    # Lowest-energy eigenstate
    ground_state = eigenvectors[:, 0]
    ground_energy = energies[0]

    # Dicke-state probabilities
    probabilities = np.abs(ground_state)**2

    # Magnetizations
    Sz_squared = Sz @ Sz

    expectation_Sz2 = np.vdot(
        ground_state,
        Sz_squared @ ground_state
    ).real

    expectation_Sx = np.vdot(
        ground_state,
        Sx @ ground_state
    ).real

    mz = 2 * np.sqrt(expectation_Sz2) / N
    mx = 2 * expectation_Sx / N

    # Identify the phase
    if np.isclose(ratio, 1.0):
        phase = "Critical region"
    elif ratio < 1.0:
        phase = "Ferromagnetic regime"
    else:
        phase = "Paramagnetic regime"

    print("=" * 50)
    print(name)
    print("=" * 50)

    print(f"N                    = {N}")
    print(f"J                    = {J}")
    print(f"Omega                = {Omega}")
    print(f"Omega/J              = {ratio:.2f}")
    print(f"Phase                 = {phase}")
    print(f"Ground-state energy   = {ground_energy:.6f}")
    print(f"z-magnetization       = {mz:.6f}")
    print(f"x-magnetization       = {mx:.6f}")

    print("\nDicke-state probabilities:")

    for k, probability in enumerate(probabilities):
        print(
            f"P(D_{k}^{N}) = {probability:.6f}"
        )

    print()
```

## Conclusion

The behavior of the LMG model is controlled by the ratio $\Omega/J$, which measures the competition between the transverse field and the collective spin interaction.

For $(J,\Omega)=(4,1)$, the ratio is $\Omega/J=0.25$. The collective interaction dominates, producing ferromagnetic ground-state behavior and small dynamical oscillations around the initial $z$-polarized state.

For $(J,\Omega)=(2,6)$, the ratio is $\Omega/J=3$. The transverse field dominates, producing paramagnetic ground-state behavior and large-amplitude spin oscillations.

The ground-state transition occurs at $\Omega/J=1$, whereas the dynamical transition for the selected initial state occurs at $\Omega/J=0.5$.

```python


N = 10
J = 1.0

cases = {
    "Ferromagnetic": 0.2,
    "Critical": 1.0,
    "Paramagnetic": 2.0
}

Sx, Sy, Sz, S_plus, S_minus = (
    im.operators.collective_spin_dicke(N)
)

Sz2 = Sz @ Sz

results = {}

for name, ratio in cases.items():

    Omega = ratio * J

    # Two competing parts
    H_interaction = -(J / N) * Sz2
    H_field = -Omega * Sx

    # Total Hamiltonian
    H = H_interaction + H_field

    # Ground state
    energies, states = np.linalg.eigh(H)
    ground_state = states[:, 0]

    # Expectation values
    mean_Sx = np.vdot(
        ground_state,
        Sx @ ground_state
    ).real

    mean_Sz2 = np.vdot(
        ground_state,
        Sz2 @ ground_state
    ).real

    # Normalized quantities
    mx = 2 * mean_Sx / N
    mz_rms = 2 * np.sqrt(mean_Sz2) / N

    probabilities = np.abs(ground_state)**2

    results[name] = probabilities

    print("=" * 50)
    print(name)
    print("=" * 50)
    print(f"Omega/J = {ratio:.1f}")
    print(f"Ground-state energy = {energies[0]:.4f}")
    print(f"z-order m_z         = {mz_rms:.4f}")
    print(f"x-alignment m_x     = {mx:.4f}")

    if ratio < 1:
        print("Conclusion: interaction along z dominates.")
    elif np.isclose(ratio, 1):
        print("Conclusion: the two effects compete strongly.")
    else:
        print("Conclusion: transverse field along x dominates.")

    print()
```

```python
J = 1.0

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

k_values = np.arange(N + 1)

for axis, (name, probabilities) in zip(
    axes,
    results.items()
):
    axis.bar(k_values, probabilities)

    axis.set_title(
        f"{name}\n"
        rf"$\Omega/J={cases[name]}$"
    )

    axis.set_xlabel(
        r"Number of spins in $|1\rangle$: $k$"
    )

    axis.set_ylabel(
        rf"$P(D_k^{{{N}}})$"
    )

    axis.set_xticks(k_values)
    axis.set_ylim(0, 0.55)
    axis.grid(axis="y", alpha=0.3)

plt.suptitle(
    "Competition Between Interaction and Transverse Field"
)

plt.tight_layout()
plt.show()
```

### Ferromagnetic state

In the ferromagnetic state, it is certain that the spins are aligned, but it may not be certain whether they are all 0 or all 1.
Consider the symmetric ferromagnetic cat state:
$$|\psi_F\rangle=\frac{|00\cdots0\rangle+|11\cdots1\rangle}{\sqrt{2}}.$$
When measured in the $z$-basis:
$$P(00\cdots0)=\left|\frac{1}{\sqrt{2}}\right|^2=\frac{1}{2},$$
and
$$P(11\cdots1)=\left|\frac{1}{\sqrt{2}}\right|^2=\frac{1}{2}.$$
Therefore:

- There is a 50% probability of measuring all zeros.
- There is a 50% probability of measuring all ones.
- There is a 100% probability of measuring spins that are all aligned.

Mathematically,
$$P(\text{aligned})=P(00\cdots0)+P(11\cdots1)=\frac{1}{2}+\frac{1}{2}=1.$$

### Example for $N=3$
$$|\psi_F\rangle=\frac{|000\rangle+|111\rangle}{\sqrt{2}}.$$
The Dicke probabilities are:
$$P(k=0)=\frac{1}{2},\qquad P(k=3)=\frac{1}{2},\qquad P(k=1)=P(k=2)=0.$$
So one measurement gives either $|000\rangle$ or $|111\rangle,$ never a mixed result such as $|001\rangle.$

### What about spontaneous symmetry breaking?
If the system selects one particular orientation, for example $|\psi\rangle=|00\cdots0\rangle,$ then $P(00\cdots0)=1.$
Alternatively, if it selects $|\psi\rangle=|11\cdots1\rangle,$ then $P(11\cdots1)=1.$

| Ferromagnetic state | Measurement probability |
|---|---|
| Symmetric cat state | $1/2$ all zeros and $1/2$ all ones |
| Symmetry-broken $+z$ state | $1$ for all zeros |
| Symmetry-broken $-z$ state | $1$ for all ones |

In your numerical graph, the two peaks near $k=0$ and $k=N$ are each approximately $0.5$. Together, they show nearly $100\%$ probability of collective ferromagnetic alignment. They are not exactly $0.5$ because your transverse field is nonzero and mixes in small contributions from other Dicke states.

## Binomial Coefficient in the Dicke Basis
The symbol
$$\boxed{\binom{N}{k}}$$
is called the **binomial coefficient** and is read as:

> “$N$ choose $k$.”

It counts the number of different ways to choose $k$ positions from a total of $N$ spins.

Its formula is
$$\boxed{\binom{N}{k}=\frac{N!}{k!(N-k)!}}$$
where $N!$ is called **$N$ factorial** and is defined as
$$N!=N(N-1)(N-2)\cdots2\cdot1.$$

### Example: $N=4$ and $k=2$
For four spins containing exactly two spins in state $|1\rangle$,
$$\binom{4}{2}=\frac{4!}{2!(4-2)!}.$$
Using $4!=24\qquad\text{and}\qquad 2!=2,$ we obtain
$$\binom{4}{2}=\frac{24}{2\times2}=6.$$
Therefore, there are six different four-spin computational-basis states containing exactly two ones:
$$|0011\rangle,\quad |0101\rangle,\quad |0110\rangle,\quad |1001\rangle,\quad |1010\rangle,\quad |1100\rangle.$$
The corresponding symmetric Dicke state is
$$|D_2^4\rangle=\frac{1}{\sqrt6}\left(|0011\rangle+|0101\rangle+|0110\rangle+|1001\rangle+|1010\rangle+|1100\rangle\right).$$
The normalization factor is $1/\sqrt6$ because the state contains six equally weighted basis states.

## Binomial Probability Distribution
For the paramagnetic state aligned along the positive $x$-direction,
$$|\psi\rangle=|+\rangle^{\otimes N},\qquad |+\rangle=\frac{|0\rangle+|1\rangle}{\sqrt2},$$
the probability of measuring exactly $k$ spins in state $|1\rangle$ is
$$\boxed{P(k)=\frac{\binom{N}{k}}{2^N}.}$$
Here, $\binom{N}{k}$ counts the number of computational-basis states containing exactly $k$ ones.

### Example: $N=10$ and $k=5$
For ten spins containing exactly five ones, $\binom{10}{5}=252.$
Therefore, there are $252$ different ten-spin strings containing five ones and five zeros.
The probability is
$$P(5)=\frac{\binom{10}{5}}{2^{10}}=\frac{252}{1024}\approx0.246.$$
This is the largest probability in the distribution because $k=5=N/2$. Therefore, the binomial distribution is centred around half of the spins being measured in state $|1\rangle$.



Using the convention that $k$ counts the number of spins in state $|1\rangle$, the symmetric Dicke states are

$$
|D_k^N\rangle,
\qquad
k=0,1,\ldots,N.
$$

The state aligned along $+x$ can be expanded as

$$
|+\rangle^{\otimes N}
=
\sum_{k=0}^{N}
\sqrt{
\frac{\binom{N}{k}}{2^N}
}
\,|D_k^N\rangle.
$$

Therefore, the probability of finding exactly $k$ spins in state $|1\rangle$ is

$$
P(k)
=
\left|
\langle D_k^N|+\rangle^{\otimes N}
\right|^2
=
\frac{\binom{N}{k}}{2^N}.
$$

This is a binomial probability distribution.

For example, for $N=3$,

$$
P(0)=\frac18,
\qquad
P(1)=\frac38,
\qquad
P(2)=\frac38,
\qquad
P(3)=\frac18.
$$

# Note That 

In the ferromagnetic regime, the Dicke probability is concentrated at $k=0$ and $k=N$, showing collective $z$-alignment. At the critical point, the probability spreads across many Dicke states because interaction and field compete. In the paramagnetic regime, it becomes a binomial distribution centred at $k=N/2$, indicating alignment along the $x$-direction.

## 6. Ground-State Critical Point

For the Hamiltonian

$$
H=-\frac{J}{N}S_z^2-\Omega S_x,
$$

the ground-state quantum critical point in the limit $N\rightarrow\infty$ is

$$
\boxed{\Omega_c=J}.
$$

Equivalently,

$$
\boxed{\frac{\Omega_c}{J}=1}.
$$

The two regimes are therefore

$$
\frac{\Omega}{J}<1
\quad\Longrightarrow\quad
\text{ferromagnetic phase},
$$

and

$$
\frac{\Omega}{J}>1
\quad\Longrightarrow\quad
\text{paramagnetic phase}.
$$

At

$$
\frac{\Omega}{J}=1,
$$

the system undergoes a ground-state quantum phase transition.

---

## 8. Order Parameters

The normalized collective magnetizations are

$$
m_x
=
\frac{2}{N}\langle S_x\rangle
$$

and

$$
m_z
=
\frac{2}{N}\sqrt{\langle S_z^2\rangle}.
$$

Their behavior is:

| Regime | $m_z$ | $m_x$ | Physical interpretation |
|---|---:|---:|---|
| $\Omega/J<1$ | Large | Smaller | Spins are ordered mainly along $z$ |
| $\Omega/J=1$ | Changes rapidly | Changes rapidly | Quantum critical region |
| $\Omega/J>1$ | Small | Approaches $1$ | Spins align mainly along $+x$ |

In the large-$N$ limit, the semiclassical order parameters are

$$
m_z=
\begin{cases}
\sqrt{1-\left(\dfrac{\Omega}{J}\right)^2},
& \Omega<J,\\[6pt]
0,
& \Omega\ge J,
\end{cases}
$$

and

$$
m_x=
\begin{cases}
\dfrac{\Omega}{J},
& \Omega<J,\\[6pt]
1,
& \Omega\ge J.
\end{cases}
$$

---

## 9. Energy Gap

The energy gap is the difference between the first excited-state energy and the ground-state energy:

$$
\Delta E=E_1-E_0.
$$

For a finite number of spins, the energy gap is generally nonzero.

As $N$ increases, the gap becomes very small near the critical point:

$$
\Omega=J.
$$

In the thermodynamic limit,

$$
N\rightarrow\infty,
$$

the appropriate excitation gap closes at the quantum critical point. This indicates that arbitrarily low-energy excitations become possible and the ground state changes non-analytically.

---

## 10. Summary

The LMG Hamiltonian

$$
H=-\frac{J}{N}S_z^2-\Omega S_x
$$

describes competition between collective spin alignment along $z$ and transverse-field alignment along $x$.

- When $\Omega/J<1$, the interaction dominates and the system is ferromagnetic.
- When $\Omega/J>1$, the transverse field dominates and the system is paramagnetic.
- At $\Omega/J=1$, the model undergoes a ground-state quantum phase transition in the large-$N$ limit.
- The quantities $m_z$, $m_x$, the Dicke-state probabilities and the energy gap can be used to identify the different regimes.

# Dynamical Quantum Phase Transition in the LMG Model

## 1. Initial State
The initial state is given by

$$|\psi_0\rangle=|S,+S\rangle,\qquad S=\frac{N}{2}.$$

In the Dicke-basis convention where $k$ counts zeros, this corresponds to

$$|\psi_0\rangle=|D_0^N\rangle=|00\cdots0\rangle,$$

with

$$S_z|\psi_0\rangle=\frac{N}{2}|\psi_0\rangle.$$

All spins initially point along $+z$, corresponding to the north pole of the collective Bloch sphere.

---

## 2. Competition Between the Terms
The Hamiltonian is

$$H=-\frac{J}{N}S_z^2-\Omega S_x.$$

The interaction term $-\frac{J}{N}S_z^2$ favours large $|S_z|$ and tries to keep the spins aligned near the $+z$ or $-z$ direction. Meanwhile, the field term $-\Omega S_x$ produces rotation about the $x$-axis, attempting to move the collective spin away from its initial $z$-polarized orientation.

| Parameter | Dynamics |
|---|---|
| $\Omega < J/2$ | The interaction dominates; the spin remains trapped near the north pole and makes small oscillations. |
| $\Omega = J/2$ | Dynamical critical point. |
| $\Omega > J/2$ | The field dominates enough for the spin to undergo large oscillations across both $z$-orientations. |

The relevant dynamical ratio and critical point are

$$\lambda_{\mathrm{dyn}}=\frac{\Omega}{J},\qquad \lambda_{\mathrm{dyn},c}=\frac{1}{2}.$$

> **Note:** This is a *dynamical* critical point. The ground-state equilibrium phase transition for this Hamiltonian occurs at $\Omega/J = 1$.

---

## 3. Time Evolution
Let the energy eigenstates satisfy

$$H|\phi_j\rangle = E_j|\phi_j\rangle.$$

Expanding the initial state in this basis yields

$$|\psi_0\rangle = \sum_j |\phi_j\rangle\langle\phi_j|\psi_0\rangle.$$

After time $t$, the state evolves to

$$|\psi(t)\rangle = e^{-iHt}|\psi_0\rangle = \sum_j e^{-iE_jt}|\phi_j\rangle\langle\phi_j|\psi_0\rangle.$$

Each energy component receives a different phase $e^{-iE_jt}$, producing the oscillatory dynamics.

---

## 4. Expectation Value of an Observable
For an observable $O$, the expectation value is

$$\langle O(t)\rangle = \langle\psi(t)|O|\psi(t)\rangle.$$

Substituting the energy expansion gives

$$\langle O(t)\rangle = \sum_{j,k} c_j^*c_k e^{-i(E_k-E_j)t} O_{jk},$$

where

$$c_j = \langle\phi_j|\psi_0\rangle,\qquad O_{jk} = \langle\phi_j|O|\phi_k\rangle.$$

- Terms with $j = k$ are constant in time.
- Terms with $j \neq k$ oscillate with frequency $\omega_{jk} = E_k - E_j$.

---

## 5. Long-Time Average
The dynamical order parameter is obtained by averaging over a long time:

$$\overline{\langle O\rangle} = \lim_{T\rightarrow\infty} \frac{1}{T}\int_0^T \langle O(t)\rangle\,dt.$$

Assuming there are no relevant degeneracies, the oscillating off-diagonal terms average to zero, leaving the diagonal ensemble:

$$\overline{\langle O\rangle} = \sum_j |c_j|^2 O_{jj}.$$

Choosing $O = S_z$, the time-averaged expectation value is

$$\overline{\langle S_z\rangle} = \lim_{T\rightarrow\infty} \frac{1}{T}\int_0^T \langle S_z(t)\rangle\,dt.$$

The normalized order parameter is defined as

$$\overline{m_z} = \frac{2}{N}\overline{\langle S_z\rangle}.$$

For large $N$, the order parameter behaves as

$$\overline{m_z} = \begin{cases} > 0, & \Omega/J < 1/2, \\[2mm] = 0, & \Omega/J > 1/2. \end{cases}$$

---

## Conclusion

- **Below $\Omega/J = 1/2$**: The collective spin remembers its initial $+z$ orientation.
- **Above $\Omega/J = 1/2$**: The spin explores both positive and negative $S_z$ regions, causing the long-time average to vanish.
- **At $\Omega/J = 1/2$**: The long-time behavior changes sharply as $N \rightarrow \infty$, marking the dynamical quantum phase transition.
